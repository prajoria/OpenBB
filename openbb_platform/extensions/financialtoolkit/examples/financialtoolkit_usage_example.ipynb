# OpenBB FinancialToolkit Extension - Usage Example\n
\n
This notebook demonstrates practical usage of the FinancialToolkit extension commands across models, risk, options, performance, and discovery.

## 1) Imports and Setup

```python
from openbb import obb\n
\n
symbols = ["AAPL", "MSFT"]\n
api_key = "YOUR_FMP_KEY"
```

## 2) Extension Health

```python
about = obb.financialtoolkit.about()\n
about
```

## 3) Models

```python
altman = obb.financialtoolkit.models.altman_z_score(symbols=symbols, api_key=api_key)\n
piotroski = obb.financialtoolkit.models.piotroski_score(symbols=symbols, api_key=api_key)\n
dupont = obb.financialtoolkit.models.dupont(symbols=symbols, api_key=api_key)\n
wacc = obb.financialtoolkit.models.wacc(symbols=symbols, api_key=api_key)\n
intrinsic = obb.financialtoolkit.models.intrinsic_value(\n
    symbols=["AAPL"],\n
    growth_rate=0.05,\n
    perpetual_growth_rate=0.025,\n
    weighted_average_cost_of_capital=0.09,\n
    api_key=api_key,\n
)\n
\n
altman.to_df().head()
```

## 4) Risk

```python
var_data = obb.financialtoolkit.risk.var(symbols=symbols, api_key=api_key, period="yearly")\n
cvar_data = obb.financialtoolkit.risk.cvar(symbols=symbols, api_key=api_key, period="yearly")\n
evar_data = obb.financialtoolkit.risk.evar(symbols=symbols, api_key=api_key, period="yearly")\n
garch_data = obb.financialtoolkit.risk.garch(symbols=symbols, api_key=api_key, period="yearly")\n
\n
var_data.to_df().head()
```

## 5) Options (Greeks)

```python
greeks = obb.financialtoolkit.options.greeks(\n
    symbols=["AAPL"],\n
    api_key=api_key,\n
    strike_price_range=0.25,\n
    strike_step_size=5,\n
    expiration_time_range=30,\n
)\n
\n
greeks.to_df().head()
```

## 6) Performance

```python
sharpe = obb.financialtoolkit.performance.sharpe_ratio(symbols=symbols, api_key=api_key, period="yearly")\n
sortino = obb.financialtoolkit.performance.sortino_ratio(symbols=symbols, api_key=api_key, period="yearly")\n
information = obb.financialtoolkit.performance.information_ratio(symbols=symbols, api_key=api_key, period="yearly")\n
\n
sharpe.to_df().head()
```

## 7) Discovery

```python
screen = obb.financialtoolkit.discovery.screen(\n
    api_key=api_key,\n
    market_cap_higher=5_000_000_000,\n
    price_higher=5,\n
    is_etf=False,\n
)\n
\n
search = obb.financialtoolkit.discovery.search(\n
    api_key=api_key,\n
    query="META",\n
    search_method="name",\n
)\n
\n
search.to_df().head()
```
